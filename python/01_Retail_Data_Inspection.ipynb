
## 5. Exploratory Data Analysis (EDA)

This section explores sales, profitability, regional performance, product performance, discount patterns, customer segments, and shipping operations to identify business trends and opportunities for improvement.



### 5.1 Initial Business Overview and Profitability

Examine shipping-mode counts and summary statistics for sales, profit, discounts, and quantities. Identify loss-making transaction records and investigate how profit varies across discount levels.


```python
df["Ship Mode"].value_counts()
```

```python
df[["Sales","Profit","Discount","Quantity"]].describe()
```

```python
df[df["Profit"]<0].shape[0]
```

```python
loss_records= df[df["Profit"]<0].shape[0]
total_records=df.shape[0]

loss_percentage=(loss_records / total_records)*100

print("Loss Records:",loss_records)
print("Total Records:", total_records)
print("Loss Percentage:", round(loss_percentage,2),"%")
```

```python
df.groupby("Discount")["Profit"].mean()
```

```python
df["Discount"].value_counts().sort_index()
```

```python
df.groupby("Discount").agg(
    Record_Count=("Profit", "count"),
    Average_Profit=("Profit","mean"),
    Total_Profit=("Profit","sum")
).round(2)
```


### 5.2 Category Performance

Compare sales and profit across product categories. Calculate profit margins to evaluate how effectively each category converts sales into profit.


```python
df.groupby("Category").agg(
    Record_Count=("Profit", "count"),
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum"),
    Average_Profit=("Profit","mean")
).round(2)
```

```python
category_summary=df.groupby("Category").agg(
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum")
)
category_summary["Profit_Margin_%"]=(
    category_summary["Total_Profit"]/category_summary["Total_Sales"]
)*100

category_summary.round(2)
```


### 5.3 Sub-Category Performance

Compare sales and total profit across product sub-categories to identify the strongest contributors to profitability and the sub-categories requiring further investigation.


```python
df.groupby("Sub-Category").agg(
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum")
).sort_values("Total_Profit").round(2)
```

### 5.4 Supporting Analysis: Tables Discount Impact

This analysis examines how discount levels relate to the profitability of Tables. The results provide product-specific supporting evidence for the overall discount and profitability analysis in Section 5.13.

```python
df[df["Sub-Category"]=="Tables"].groupby("Discount").agg(
    Record_Count=("Profit","count"),
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum"),
    Average_Profit=("Profit","mean")
).round(2)
```

```python
tables_summary = df[df["Sub-Category"] == "Tables"].groupby("Discount").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

tables_summary["Profit_Margin_%"] = (
    tables_summary["Total_Profit"] / tables_summary["Total_Sales"]
) * 100

tables_summary.round(2)
```

### 5.5 Regional Performance

Compare sales, profit, profit margin, record count, and quantity across regions to identify strong-performing markets and regions that require profitability improvements.

```python
regional_summary = df.groupby("Region").agg(
    Record_Count=("Profit", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
)

regional_summary["Profit_Margin_%"] = (
    regional_summary["Total_Profit"] /
    regional_summary["Total_Sales"]
) * 100

regional_summary.sort_values(
    "Total_Profit", ascending=False
).round(2)
```

### 5.6 Central Region Sub-Category Profitability

Analyze individual sub-categories within the Central region to identify specific sources of profit and loss. This analysis provides a deeper level of detail than the regional category comparison in Section 5.18.

```python
central_summary=df[df["Region"]== "Central"].groupby("Sub-Category").agg(
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum")
)

central_summary["Progit_Margin_%"]=(
    central_summary["Total_Profit"]/central_summary["Total_Sales"]
)*100

central_summary.round(2).sort_values("Total_Profit")
```

### 5.7 Supporting Analysis: Central Furnishings Discount Impact

Examine how discount levels relate to Furnishings profitability in the Central region. Use the results as a focused example supporting the overall discount analysis in Section 5.13 and the regional comparison in Section 5.17.

```python
central_Furnishings=df[
    (df["Region"]=="Central") &
    (df["Sub-Category"]=="Furnishings")
]

central_Furnishings.groupby("Discount").agg(
    Record_Count=("Profit","count"),
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum")
).round(2)
```

### 5.8 Supporting Analysis: Product-Level Losses

Investigate individual Central-region Furnishings transactions receiving a 60% discount to identify products associated with significant losses. Use the results as transaction-level evidence supporting the discount and profitability findings in Section 5.13.

```python
central_furnishings_60 = df[
    (df["Region"] == "Central") &
    (df["Sub-Category"] == "Furnishings") &
    (df["Discount"] == 0.60)
]
```

```python
central_furnishings_60.groupby("Product Name").agg(
    Record_Count=("Profit", "count"),
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum")
).sort_values("Total_Profit").round(2)
```

```python
product_summary = central_furnishings_60.groupby("Product Name").agg(
    Record_Count=("Profit", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

product_summary["Profit_Margin_%"] = (
    product_summary["Total_Profit"] /
    product_summary["Total_Sales"]
) * 100

product_summary.round(2).sort_values("Profit_Margin_%")
```

```python
product_summary.sort_values("Total_Profit").head(5)
```

### 5.9 Customer Segment Profitability Analysis

Compare sales, profit, order volume, and profit margin across customer segments to identify which segments contribute most to business performance and which may require further investigation.

```python
segment_summary=df.groupby("Segment").agg(
    Record_count=("Order ID","count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit","sum"),
    Total_Quantity=("Quantity","sum"),
    Average_Discount=("Discount","mean")
)

segment_summary["Profit_Margin_%"]=(
    segment_summary["Total_Profit"]
    /segment_summary["Total_Sales"]
)*100

segment_summary.round(2).sort_values(
    "Total_Profit", ascending=False
)
```

### 5.10 Shipping Mode Profitability Analysis

Compare sales, profit, quantity, and profit margin across shipping modes to understand their contribution to business performance and identify shipping modes that may require further investigation.

```python
df["Ship Mode"].value_counts()
```

```python
shipping_summary=df.groupby("Ship Mode").agg(
    Record_count=("Order ID","count"),
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum"),
    Total_Quantity=("Quantity","sum"),
    Average_Discount=("Discount","mean")
)

shipping_summary["Profit_Margin_%"]=(
    shipping_summary["Total_Profit"]
    / shipping_summary["Total_Sales"]
)*100

shipping_summary.round(2).sort_values(
    "Total_Profit", ascending=False
)
```

### 5.11 Shipping Days Analysis

Analyze shipping duration across different shipping modes to identify delivery-time differences and opportunities to improve delivery performance.

```python
shipping_days_summary=df.groupby("Ship Mode").agg(
    Record_count=("Order ID","count"),
    Average_Shipping_Days=("Shipping Days","mean"),
    Median_Shipping_Days=("Shipping Days","median"),
    Minimum_Shipping_Days=("Shipping Days","min"),
    Maximum_Shipping_Days= ("Shipping Days","max")

)

shipping_days_summary.round(2).sort_values(
    "Average_Shipping_Days"
)
```

### 5.12 Discount and Profitability Analysis

Analyze sales, total profit, and profit margin across different discount levels to understand how discounting relates to business profitability.



```python
discount_summary=df.groupby("Discount").agg(
    Record_Count=("Profit","count"),
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum")
)

discount_summary["Profit_Margin_%"]=(
    discount_summary["Total_Profit"]
    /discount_summary["Total_Sales"]
)*100

discount_summary.round(2).sort_index()
```

### 5.13 Sub-Category Profitability Analysis

Compare sales, profit, quantity sold, and profit margin across product sub-categories to identify which sub-categories contribute most to business performance and which may require further investigation.

```python
subcategory_summary = df.groupby("Sub-Category").agg(
    Record_Count=("Order ID", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
)

subcategory_summary["Profit_Margin_%"] = (
    subcategory_summary["Total_Profit"]
    / subcategory_summary["Total_Sales"]
) * 100

subcategory_summary.round(2).sort_values(
    "Total_Profit", ascending=False
)
```

### 5.14 State-Level Profitability Analysis

Compare sales, profit, order volume, and profit margin across states to identify which states contribute most to business performance and which may require further investigation.

```python
state_summary = df.groupby("State").agg(
    Record_Count=("Order ID", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
)

state_summary["Profit_Margin_%"] = (
    state_summary["Total_Profit"]
    / state_summary["Total_Sales"]
) * 100

state_summary.round(2).sort_values(
    "Total_Profit", ascending=False
)
```

### 5.15 Supporting Analysis: Regional Discount Profitability

Compare profitability across discount levels within each region. Use these results to add regional context to the overall discount and profitability analysis in Section 5.13, with particular attention to discount levels associated with losses.

```python
regional_discount_summary = df.groupby(
    ["Region", "Discount"]
).agg(
    Record_Count=("Profit", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

regional_discount_summary["Profit_Margin_%"] = (
    regional_discount_summary["Total_Profit"]
    / regional_discount_summary["Total_Sales"]
) * 100

regional_discount_summary.round(2).sort_values(
    ["Region", "Discount"]
)
```

### 5.16 Product Category Profitability by Region

Compare sales, total profit, and profit margin across product categories and regions to identify which category-region combinations contribute most to business performance and which may require further investigation.

```python
category_region_summary = df.groupby(
    ["Region", "Category"]
).agg(
    Record_Count=("Order ID", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

category_region_summary["Profit_Margin_%"] = (
    category_region_summary["Total_Profit"]
    / category_region_summary["Total_Sales"]
) * 100

category_region_summary.round(2).sort_values(
    ["Region", "Total_Profit"],
    ascending=[True, False]
)
```

### 5.17 Supporting Analysis: Category and Discount Profitability

Examine how profitability varies across discount levels within Furniture, Office Supplies, and Technology. Use the results to identify category-specific loss patterns that support the overall discount analysis in Section 5.13.

```python
category_discount_summary = df.groupby(
    ["Category", "Discount"]
).agg(
    Record_Count=("Order ID", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

category_discount_summary["Profit_Margin_%"] = (
    category_discount_summary["Total_Profit"]
    / category_discount_summary["Total_Sales"]
) * 100

category_discount_summary.round(2).sort_values(
    ["Category", "Discount"]
)
```

### 5.18 Monthly Sales and Profit Trends

Analyze monthly sales and profit over time to identify seasonal patterns, changes in business performance, and months that may require further investigation.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_summary = df.groupby(
    df["Order Date"].dt.to_period("M")
).agg(
    Record_Count=("Order ID", "count"),
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

monthly_summary.round(2)
```

# 6. Key Findings and Business Recommendations

Document the most important findings from the analysis and translate them into practical business recommendations.

## 6.1 Discount and Profitability

- **Business Question:** How do discount levels relate to profitability?
- **Evidence:** At 0% discount, profit was 320,987.60 with a 29.51% margin. At 70% discount, the business recorded a loss of 40,075.36 with a -98.66% margin.
- **Interpretation:** Higher discounts are associated with losses in several discount groups.
- **Recommendation:** Review high-discount transactions and establish discount limits based on product profitability.

## 6.2 Regional Profitability

- **Business Question:** Which regions perform best and worst in terms of profitability?
- **Evidence:** West generated the highest profit of 108,418.45, while Central had the lowest profit margin at 7.92%.
- **Interpretation:** Profitability differs across regions, suggesting opportunities to improve weaker regional performance.
- **Recommendation:** Investigate product mix, discounting, and loss-making transactions in the Central region.

## 6.3 Product and Sub-Category Performance

- **Business Question:** Which sub-categories contribute most to profit and loss?
- **Evidence:** Copiers generated the highest total profit of 55,617.82, while Tables recorded the largest loss of 17,725.48.
- **Interpretation:** Product performance varies considerably, and some sub-categories may require pricing or discount reviews.
- **Recommendation:** Review pricing, costs, and discount policies for loss-making sub-categories, especially Tables and Bookcases.

## 6.4 Category Profitability Across Regions

- **Business Question:** Does category profitability vary across regions?
- **Evidence:** Central Furniture recorded a loss of 2,871.05, while West Office Supplies generated a profit of 52,609.85.
- **Interpretation:** The same product category can perform differently across regions.
- **Recommendation:** Investigate regional pricing, product demand, and discount patterns before making region-specific decisions.

## 6.5 Customer Segment Performance

- **Business Question:** Which customer segment contributes the most profit?
- **Evidence:** Consumer generated the highest total profit of 134,119.21, while Home Office had the highest profit margin at 14.03%.
- **Interpretation:** Consumer contributes the most overall profit, while Home Office generates a higher return on sales.
- **Recommendation:** Maintain engagement with Consumer customers and investigate opportunities to grow profitable Home Office sales.

## 6.6 Shipping and Operational Performance

- **Business Question:** How do shipping modes differ in profitability and delivery time?
- **Evidence:** Standard Class generated the highest total profit of 164,088.79, while First Class had the highest profit margin at 13.93%. Average shipping time for Standard Class was 5.01 days.
- **Interpretation:** Shipping modes differ in sales volume, profitability, and delivery speed.
- **Recommendation:** Monitor delivery performance and profitability together when reviewing shipping options.

## 6.7 Monthly Sales and Profit Trends

- **Business Question:** How do sales and profit change over time?
- **Evidence:** Monthly sales and profit fluctuated across the 2014–2017 period, with sales reaching 118,447.82 in November 2017.
- **Interpretation:** Monthly performance varies, so sales growth does not necessarily mean profit grows at the same rate.
- **Recommendation:** Monitor monthly sales, profit, and profit margin to identify unusual changes and investigate their possible causes.
